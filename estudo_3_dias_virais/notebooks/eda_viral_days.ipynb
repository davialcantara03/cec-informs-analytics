```python
import pandas as pd
df_cec = pd.read_csv(r'C:\Users\davim\OneDrive\Documentos\projeto_3_cec_informs\data\dataset_cecinforms.csv', sep=';', encoding='utf-8')
df_cec.head()
```

```python
# 1. Calculando Média e Desvio-Padrão da coluna 'impressoes'
media_impressoes = df_cec['impressoes'].mean()
desvio_padrao = df_cec['impressoes'].std()

# 2. Linha de corte do "Dia Monstro" (Média + 2 Desvios-Padrão)
limite_dia_monstro = media_impressoes + (2 * desvio_padrao)

# Exibindo os valores de referência estatística
print(f"📊 Média diária de impressões: {media_impressoes:,.0f}".replace(",", "."))
print(f"📏 Desvio-Padrão: {desvio_padrao:,.0f}".replace(",", "."))
 .replace(",", "."))

# 3. Filtrando a tabela com os Dias Monstro
dias_monstro = df_cec[df_cec['impressoes'] > limite_dia_monstro]

print(f"⚡ Total de 'Dias Monstro' encontrados: {len(dias_monstro)}")

# 4. Usando o display() para renderizar a tabela formatada no Jupyter
display(dias_monstro.head(16))
```

```python
# 1. Maiores picos absolutos de impressões do ano
print("🏆 TOP 5 MAIORES DIAS DO ANO:")
display(dias_monstro.sort_values(by='impressoes', ascending=False)[['data', 'dia_semana', 'impressoes', 'engajamentos', 'novos_seguidores']].head(5))

# 2. Distribuição por Dia da Semana (onde a viralidade mais acontece?)
print("\n📅 DISTRIBUIÇÃO DOS DIAS MONSTRO POR DIA DA SEMANA:")
contagem_dias = dias_monstro['dia_semana'].value_counts().reset_index()
contagem_dias.columns = ['Dia da Semana', 'Quantidade de Picos']
display(contagem_dias)
```

```python
# 1. Garantir que a coluna de data está no formato correto e ordenada
df_cec['data'] = pd.to_datetime(df_cec['data'])
df_cec = df_cec.sort_values('data').reset_index(drop=True)

# 2. Identificar os índices dos Dias Monstro
indices_monstro = df_cec[df_cec['impressoes'] > limite_dia_monstro].index

# 3. Mapear a janela de ±3 dias ao redor de cada pico
janelas = []
for idx in indices_monstro:
    inicio = max(0, idx - 3)
    fim = min(len(df_cec), idx + 4)
    trecho = df_cec.iloc[inicio:fim].copy()
    trecho['offset_dias'] = range(inicio - idx, fim - idx)
    janelas.append(trecho)

df_arrasto = pd.concat(janelas)

# 4. Agrupar as médias e formatar a exibição
tabela_arrasto = df_arrasto.groupby('offset_dias').agg(
    media_impressoes=('impressoes', 'mean'),
    media_engajamentos=('engajamentos', 'mean'),
    media_seguidores=('novos_seguidores', 'mean')
).reset_index()

# 5. Mapear nomes amigáveis para as posições temporais
tabela_arrasto['Posicao'] = tabela_arrasto['offset_dias'].map({
    -3: '3 Dias Antes',
    -2: '2 Dias Antes',
    -1: '1 Dia Antes',
     0: 'DIA DO PICO',
     1: '1 Dia Depois (+1)',
     2: '2 Dias Depois (+2)',
     3: '3 Dias Depois (+3)'
})

display(tabela_arrasto[['Posicao', 'media_impressoes', 'media_engajamentos', 'media_seguidores']])
```

```python
# Formatando para exibição sem notação científica
tabela_arrasto_formatada = tabela_arrasto.copy()
tabela_arrasto_formatada['media_impressoes'] = tabela_arrasto_formatada['media_impressoes'].round(0)
tabela_arrasto_formatada['media_engajamentos'] = tabela_arrasto_formatada['media_engajamentos'].round(0)
tabela_arrasto_formatada['media_seguidores'] = tabela_arrasto_formatada['media_seguidores'].round(1)

display(tabela_arrasto_formatada)
```

```python
# 1. Criando métricas relativas (taxas)
df_cec['taxa_engajamento'] = (df_cec['engajamentos'] / df_cec['impressoes']) * 100
df_cec['taxa_conversao_seguidores'] = (df_cec['novos_seguidores'] / df_cec['impressoes']) * 1000  # Seguidores a cada 1k impressões

# 2. Separando Dias Monstro de Dias Normais
dias_monstro = df_cec[df_cec['impressoes'] > limite_dia_monstro]
dias_normais = df_cec[df_cec['impressoes'] <= limite_dia_monstro]

# 3. Comparando os dois cenários
comparativo = pd.DataFrame({
    'Métrica': ['Impressões Médias', 'Engajamento Médio', 'Taxa de Engajamento (%)', 'Seguidores por 1k Impressões'],
    'Dias Normais': [
        dias_normais['impressoes'].mean(),
        dias_normais['engajamentos'].mean(),
        dias_normais['taxa_engajamento'].mean(),
        dias_normais['taxa_conversao_seguidores'].mean()
    ],
    'Dias Monstro': [
        dias_monstro['impressoes'].mean(),
        dias_monstro['engajamentos'].mean(),
        dias_monstro['taxa_engajamento'].mean(),
        dias_monstro['taxa_conversao_seguidores'].mean()
    ]
})

display(comparativo.round(2))
```
